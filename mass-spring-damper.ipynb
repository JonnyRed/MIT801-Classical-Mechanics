```python
import sympy as sp
import sympy.physics.mechanics as me
from sympy.physics. vector.printing import init_vprinting

import IPython
from IPython.display import display, Math, Markdown

sp.init_printing()
init_vprinting(use_latex='mathjax')
```

```python
def reference_frame(frame: str, x=r"\imath", y=r"\jmath", z=r"k") -> me.ReferenceFrame:
    """Create a SymPy reference frame with custom basis vector labels.

    Parameters
    ----------
    frame : str
        The name of the reference frame.
    x, y, z : str
        Labels for the basis vectors.
    """
    return me.ReferenceFrame(
        frame,
        latexs=(
            rf"\; {{}}^\mathcal {frame} \hat {x}",
            rf"\;{{}}^\mathcal {frame} \hat {y}",
            rf"\: {{}}^\mathcal {frame} \hat {z}",
        ),
    )


def reference_frame_circular(name: str, angle=r"theta") -> me.ReferenceFrame:
    """Create a circular reference frame with radial and angular basis labels.

    Parameters
    ----------
    name : str
        Name of the new reference frame.
    angle : str, optional
        Symbol or label used for the angular basis vector, by default "theta".
    """
    return reference_frame(name, x=r"r", y=rf"\{angle}", z=r"e_z")
```

* 𝑥⁡(𝑡): distance of the particle from the ceiling
* 𝑣⁡(𝑡): speed of the particle
* 𝑚: mass of the particle
* 𝑐: damping coefficient of the damper
* 𝑘: stiffness of the spring
* 𝑔: acceleration due to gravity
* 𝑡: time

```python
(
    x, # distance of the particle from the ceiling
    v  # speed  of the particle
) = me.dynamicsymbols('x v')

(
    m, # mass of the particle
    c, # damping coefficient
    k, # spring stiffness
    g, # gravitational acceleration
    t  # time
) = sp.symbols('m c k g t')

# Newtonian reference frame that represents the ceiling which the 
# particle is attached to, 𝐶

N = reference_frame('N')
```

```python
# We will need two points, one to represent the original position of 
# the particle which stays fixed in the ceiling frame, 𝑂, and the 
# second one, 𝑃 which is aligned with the particle as it moves.

O = me.Point('O')
O.set_vel(N, 0) # velocity of the ceiling point is zero in the ceiling frame

P = me.Point('P')
P.set_pos(O, x * N.x)
P.set_vel(N, v * N.x)

# velocity of the particle in the ceiling frame
display(Math(r'\vec{v}_{P/N} = ' + sp.latex(P.vel(N))))
```

```python
# There are three forces acting on the particle. Those due to the acceleration 
# of gravity, the damper, and the spring.

damping = -c * P.vel(N)
stiffness = -k * P.pos_from(O)
gravity = m * g * N.x
forces = damping + stiffness + gravity

display(Math(r'\vec{F}_{P/N} = ' + sp.latex(forces)))

# Now we can use Newton’s second law, 0 =𝐹 −𝑚⁢𝑎, to form the equation of motion 
# of the system.

N2L_spring_damper = sp.Eq(forces.to_matrix(N), m * P.acc(N).to_matrix(N))
display(Math(r'\text{Equation of motion N2L: }' + sp.latex(N2L_spring_damper)))

sol_spring_damper = sp.solve(N2L_spring_damper, v.diff(t))
vel_spring_damper = sol_spring_damper[v.diff(t)].cancel()
display(
    Math(
        r'\text{acceleration of the particle: }' 
        + me.vlatex(v.diff(t)) + " = " 
        + sp.latex(vel_spring_damper)
    )
)
```
