Mapping them out in dimensionless form creates a powerful reference toolkit.

When analyzing constant acceleration ($a$) where there is a non-zero initial velocity ($v_0$), we use the standard reference scales:

* **Velocity Scale:** $V_c = v_0 \implies v^* = \frac{v}{v_0}$
* **Time Scale:** $T_c = \frac{v_0}{a} \implies t^* = \frac{at}{v_0}$
* **Length Scale:** $L_c = \frac{v_0^2}{a} \implies s^* = \frac{as}{v_0^2}$

### Table of Dimensionless Kinematic Equations

| Physical Kinematic Equation | Description | Dimensionless Form |
|---|---|---|
| $v = v_0 + at$ | Velocity–Time | $v^* = 1 + t^*$ |
| $s = v_0 t + \frac{1}{2}at^2$ | Displacement–Time | $s^* = t^* + \frac{1}{2}(t^*)^2$ |
| $v^2 = v_0^2 + 2as$ | Velocity–Displacement | $(v^*)^2 = 1 + 2s^*$ |
| $s = \frac{v + v_0}{2} t$ | Average Velocity Displacement | $s^* = \frac{v^* + 1}{2} t^*$ |

## Real Values Example: v₀ = 2 m/s, a = g = 9.81 m/s²


```python
import matplotlib.pyplot as plt
import numpy as np

# Define parameters
v0 = 2        # m/s (initial velocity)
g = 9.81      # m/s² (gravitational acceleration)

# Define dimensionless time t* from 0 to 4
t_star = np.linspace(0, 4, 100)

# Dimensionless equations
v_star = 1 + t_star # v* = 1 + t*
s_star = t_star + 0.5 * t_star**2 # s* = t* + 0.5*(t*)^2

# Convert to real values
t_scale = v0 / g
t = t_star * t_scale           # Real time in seconds
v = v_star * v0                # Real velocity in m/s
s = s_star * (v0**2 / g)       # Real displacement in meters

# Side-by-side comparison
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 10))

# Dimensionless plot
ax1.plot(t_star, v_star, label=r'$v^* = 1 + t^*$', color='teal', lw=2)
ax1.plot(t_star, s_star, label=r'$s^* = t^* + \frac{1}{2}(t^*)^2$', color='coral', lw=2)
ax1.set_xlabel(r'Dimensionless Time ($t^*$)')
ax1.set_ylabel(r'Dimensionless Values')
ax1.set_title('Dimensionless Form')
ax1.grid(True, linestyle='--', alpha=0.6)
ax1.legend()

# Real values plot
ax2.plot(t, v, label='Velocity', color='teal', lw=2)
ax2.plot(t, s, label='Displacement', color='coral', lw=2)
ax2.set_xlabel('Time (s)')
ax2.set_ylabel('Real Values (m/s, m)')
ax2.set_title(f'Real Values (v₀={v0} m/s, a=g={g} m/s²)')
ax2.grid(True, linestyle='--', alpha=0.6)
ax2.legend()

plt.tight_layout()
plt.show()

# Print scaling factors for reference
print(f"Scaling factors for v₀ = {v0} m/s and a = {g} m/s²:")
print(f"Time scale: v₀/a = {t_scale:.4f} s")
print(f"Velocity scale: v₀ = {v0} m/s")
print(f"Displacement scale: v₀²/a = {v0**2/g:.4f} m")
print(f"\nMaximum values (at t* = 4):")
print(f"Max time: {t[-1]:.4f} s")
print(f"Max velocity: {v[-1]:.4f} m/s")
print(f"Max displacement: {s[-1]:.4f} m")
```
